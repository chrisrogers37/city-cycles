```python
import psycopg2
import pandas as pd
import dotenv
import os
```

```python
dotenv.load_dotenv()
```

```python
mart_tables = [
'mart_nyc_hourly_patterns',
'mart_london_daily_metrics',
'mart_daily_metrics_long', 
'mart_london_station_growth',
'mart_nyc_member_analysis',
'mart_nyc_station_growth',
'mart_london_hourly_patterns',
'mart_nyc_daily_metrics'
]     
```

```python
schema = 'dbt_models_marts'
```

```python
conn = psycopg2.connect(
    dbname=os.environ['DB_NAME'],
    user=os.environ['DB_USER'],
    password=os.environ['DB_PASSWORD'],
    host=os.environ['DB_HOST'],
    port=os.environ['DB_PORT']
)

for table in mart_tables:
    print(f"Querying table: {table}")
    query = f"SELECT * FROM {schema}.{table}"  # example mart
    print(f"Query: {query}")
    df = pd.read_sql(query, conn)
    print("Table returned as pandas df.")
    df.to_parquet(f"{table}.parquet", engine="pyarrow", index=False)
    print(f"Pandas df saved as parquet to: ~/data/{table}.parquet")
    print("\n")
```
